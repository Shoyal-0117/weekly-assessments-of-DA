## Q1. Ans:

A Series is a one-dimensional labeled array it's like a single column, while a DataFrame is a two-dimensional table made of multiple Series.
Operations like .mean() work on both. Operations like selecting multiple columns (df[['col1','col2']]) work only on DataFrames.

## Q2. Ans:

inplace=True modifies the original DataFrame directly instead of returning a new one.

Pandas discourages it because it can make code harder to debug and doesn’t always improve performance.

## Q3. Ans:

.fillna() replaces missing values with a chosen value, while .dropna() removes rows or columns containing missing values.
Use .fillna() when missing data is still useful like replacing missing age with average age. Use .dropna() when missing values make rows unusable.


## Q4. Ans:

errors='coerce' converts invalid values into NaN instead of throwing an error.
Plain pd.to_numeric() raises an error if bad values exist. For messy real-world data, errors='coerce' is better because it handles invalid entries safely.

## Q5. Ans:


.loc selects rows/columns using labels, while .iloc uses integer positions.

Example:

- df.loc[2] → row with label 2
- df.iloc[2] → third row

They differ when index labels are not sequential. If index is [10,20,30], df.loc[0] fails but df.iloc[0] returns the first row.


## Q6. Ans:

- .map() maps each value using a dictionary or function.
- .apply() applies a custom function to each element.
- .replace() directly substitutes specific values.

Use .map() for category mapping, .apply() for custom logic, and .replace() for simple value replacement.

## Q7. Ans:

The SettingWithCopyWarning is a pandas warning that alerts you when you attempt to assign values to what could be either a view or a copy of an existing DataFrame.




```
import pandas as pd

df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})

# This triggers the warning
filtered = df[df['A'] > 1]
filtered['B'] = 99

# Recommended fix
df.loc[df['A'] > 1, 'B'] = 99

# Or explicitly make a copy
filtered = df[df['A'] > 1].copy()
filtered['B'] = 99

```

Using .loc[] ensures you're clearly assigning on the original DataFrame, while .copy() makes it explicit that you're working on an independent subset—both approaches eliminate ambiguity and prevent the warning.









## Q8. Ans:
pd.cut() — Equal Width Binning
Creates bins of equal range across the data spread. Each bin covers the same interval width regardless of how many data points fall within it.

```
# Example
scores = [45, 52, 60, 78, 82, 95]
bins = pd.cut(scores, bins=3)  # Divides range 45-95 into 3 equal-width ranges
```
pd.qcut() — Equal Frequency Binning (Quantile-Based)
Creates bins that contain roughly equal numbers of observations. Each bin represents an equal proportion of the dataset.

```
# Example  
income = [10, 15, 20, 25, 30, 1000]  # Skewed distribution
qs = pd.qcut(income, q=3)  # Creates 3 buckets with ~33% of data each
```

# Section C – Applied on Indian Startups Dataset

## Loading & Inspection


### Q17. Ans:


````python
# the section C is ai assested :
# for logic checking and same logice building


import pandas as pd
import numpy as np

df = pd.read_csv("/content/indian_startups_fundingDADS.csv")

# Clean column names
df.columns = [
    'startup_name',
    'city',
    'industry',
    'funding_stage',
    'amount',
    'funding_date',
    'investors',
    'founded_year'
]

# Strip + title case
df['startup_name'] = df['startup_name'].str.strip().str.title()
df['city'] = df['city'].str.strip().str.title()

# City aliases
city_map = {
    'Bangalore': 'Bengaluru',
    'Blr': 'Bengaluru',
    'Bombay': 'Mumbai',
    'Gurgaon': 'Gurugram',
    'Delhi': 'New Delhi',
    'Calcutta': 'Kolkata',
    'Madras': 'Chennai'
}
df['city'] = df['city'].replace(city_map)

# Industry cleaning
industry_map = {
    'Fin Tech': 'Fintech',
    'Fintech ': 'Fintech',
    'Ed Tech': 'Edtech',
    'Health Tech': 'HealthTech'
}
df['industry'] = df['industry'].replace(industry_map)

# Funding stage cleaning
stage_map = {
    'Seed Round': 'Seed',
    'Series A Round': 'Series A',
    'Series B Round': 'Series B'
}
df['funding_stage'] = df['funding_stage'].replace(stage_map)

# Amount cleaning
df['amount'] = (
    df['amount']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace('INR', '', regex=False)
    .str.replace('Cr', '', regex=False)
    .str.replace('Crore', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
)
df['amount'] = df['amount'].replace('Undisclosed', np.nan)
df['amount'] = pd.to_numeric(df['amount'], errors='coerce')

# Date cleaning
df['funding_date'] = pd.to_datetime(
    df['funding_date'],
    errors='coerce',
    dayfirst=True
)

# Remove duplicates
df = df.drop_duplicates()
````
