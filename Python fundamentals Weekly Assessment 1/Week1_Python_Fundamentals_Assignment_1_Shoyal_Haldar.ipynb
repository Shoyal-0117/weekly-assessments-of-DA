```python
annual_income = float(input("Enter your annual income: "))

slab1_tax = 0
slab2_tax = 0
slab3_tax = 0
slab4_tax = 0
slab5_tax = 0
slab6_tax = 0

# Slab 1: 0 - 3,00,000 → 0%
# no tax

# Slab 2: 3,00,001 - 7,00,000 → 5%
if annual_income > 300000:
    slab2_tax = (min(annual_income, 700000) - 300000) * 0.05

# Slab 3: 7,00,001 - 10,00,000 → 10%
if annual_income > 700000:
    slab3_tax = (min(annual_income, 1000000) - 700000) * 0.10

# Slab 4: 10,00,001 - 12,00,000 → 15%
if annual_income > 1000000:
    slab4_tax = (min(annual_income, 1200000) - 1000000) * 0.15

# Slab 5: 12,00,001 - 15,00,000 → 20%
if annual_income > 1200000:
    slab5_tax = (min(annual_income, 1500000) - 1200000) * 0.20

# Slab 6: 15,00,001+ → 30%
if annual_income > 1500000:
    slab6_tax = (annual_income - 1500000) * 0.30

total_tax = slab1_tax + slab2_tax + slab3_tax + slab4_tax + slab5_tax + slab6_tax

print(f"Slab 1 tax (0-3L)     : ₹{slab1_tax}")
print(f"Slab 2 tax (3L-7L)    : ₹{slab2_tax}")
print(f"Slab 3 tax (7L-10L)   : ₹{slab3_tax}")
print(f"Slab 4 tax (10L-12L)  : ₹{slab4_tax}")
print(f"Slab 5 tax (12L-15L)  : ₹{slab5_tax}")
print(f"Slab 6 tax (15L+)     : ₹{slab6_tax}")
print(f"Total tax payable     : ₹{total_tax}")
```

```python
result = []
for num in range(1, 10000):
    digits = len(str(num))
    total = 0
    temp = num
    while temp > 0:
        total += (temp % 10) ** digits
        temp //= 10
    if total == num:
        result.append(num)
print(", ".join(map(str, result)))
print(f"Count: {len(result)}")
```

### Q33.Ans:


```python
n = int(input("Enter N: "))

def is_prime(num):
    if num < 2:
        return False
    i = 2
    while i * i <= num:  # sqrt tak — no library!
        if num % i == 0:
            return False
        i += 1
    return True

# check if n itself is prime
print(f"{n} is {'Prime' if is_prime(n) else 'Not Prime'}")

# sum of all primes from 2 to n
prime_sum = 0
for num in range(2, n+1):
    if is_prime(num):
        prime_sum += num

print(f"Sum of all primes from 2 to {n}: {prime_sum}")
```

## C5 – Lists, Tuples, Sets, Dicts (Q37-Q42)

### Q37.Ans:


```python
while True:
    marks = list(map(int, input("Enter exactly 10 marks: ").split()))
    if len(marks) == 10:
        break
    print(f"Error: You entered {len(marks)} marks. Please enter exactly 10!")

max_m = marks[0]
min_m = marks[0]
total = 0
for m in marks:
    if m > max_m: max_m = m
    if m < min_m: min_m = m
    total += m
avg = total / len(marks)
above_avg = sum(1 for m in marks if m > avg)

# second highest
unique = list(set(marks))
unique_sorted_max = unique[0]
second = unique[0]
for u in unique:
    if u > unique_sorted_max:
        second = unique_sorted_max
        unique_sorted_max = u
    elif u > second and u != unique_sorted_max:
        second = u

print(f"""
Max        : {max_m}
Min        : {min_m}
Average    : {avg:.2f}
Above avg  : {above_avg}
2nd highest: {second}""")
```

```python
prices = (3845, 3902, 3878, 3955, 4012, 3998, 4055)
days = ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]

# here max(prices) will give the max value
# then i used that max value and find its index i, becouse the ith day in the days list will be the max value day so the {days[prices.index(max(prices))]} will retrun the day
print(f"Max: {max(prices)} on {days[prices.index(max(prices))]}")
print(f"Min: {min(prices)} on {days[prices.index(min(prices))]}")
print(f"Average: {sum(prices)/len(prices):.2f}")

max_gain = 0
gain_days = ""
for i in range(1, len(prices)):
    gain = prices[i] - prices[i-1]
    if gain > max_gain:
        max_gain = gain
        gain_days = f"{days[i-1]} to {days[i]}"
print(f"Biggest gain: ₹{max_gain} ({gain_days})")

volatility = sum(abs(prices[i]-prices[i-1]) for i in range(1,len(prices)))
print(f"Volatility: ₹{volatility}")
```

```python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    if gst_inclusive:
        base = round(amount / (1 + gst_rate/100), 2)
        gst = round(amount - base, 2)
    else:
        base = round(amount, 2)
        gst = round(amount * gst_rate/100, 2)
    total = round(base + gst, 2)
    return (base, gst, total)

print(calculate_gst(1000))
print(calculate_gst(1180, gst_inclusive=True))
print(calculate_gst(500, gst_rate=5))
```

### Q47.Ans:


```python
def power(base, exp):
    if exp == 0:
        return 1
    if exp < 0:
        return round(1 / power(base, -exp),4)
    return round(base * power(base, exp-1),4)

print(power(2,10))
print(power(5,0))
print(power(3,-2))
```

### Q48.Ans:


```python
import numpy as np
identity_mat = np.eye(5)
print("Identity Matrix: ")
print(identity_mat)
print()

mat = identity_mat * 10

print("After Multipication with 10 the matrix become: ")
print(mat)
print()

print("After Transpose: ")
print(mat.T)  # transpose = same ,becouse mat has same digonal elements
print()

print("Sum of all the digonal elements: ",np.trace(mat))   # diagonal sum
print("Sum of all the elements in the matrix: ",np.sum(mat))     # all elements sum
```

### Q49.Ans:


```python
import numpy as np
sales = np.array([12000,18500,14200,22100,19800,35200,28400])
days = ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]
print(f"Total: ₹{np.sum(sales)}")
print(f"Average: ₹{np.mean(sales):.2f}")
print(f"Max day: {days[np.argmax(sales)]}")
print(f"Above average days: {np.sum(sales > np.mean(sales))}")
print(f"Growth Mon-Sun: {((sales[-1]-sales[0])/sales[0])*100:.2f}%")
changes = np.diff(sales) / sales[:-1] * 100
print(f"Day-on-day changes: {np.round(changes, 2)}")
```
